```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

geolocator = Nominatim(user_agent="hyderabad_water_project", timeout=10)

geocode = RateLimiter(
    geolocator.geocode,
    min_delay_seconds=1,
    max_retries=2,
    error_wait_seconds=2.0,
    swallow_exceptions=True
)

section_coords = []
total = len(df2["section"].unique())

for i, section in enumerate(df2["section"].unique(), start=1):
    query = f"{section}, Hyderabad, India"
    location = geocode(query)
    if location:
        print(f"[{i}/{total}] {section} -> ({location.latitude}, {location.longitude})")
        section_coords.append({"section": section, "latitude": location.latitude, "longitude": location.longitude})
    else:
        print(f"[{i}/{total}] {section} -> NOT FOUND")
        section_coords.append({"section": section, "latitude": None, "longitude": None})

section_coords_df = pd.DataFrame(section_coords)
print("\nDone.")
```

```python
not_found = section_coords_df[section_coords_df["latitude"].isna()]
print(f"{len(not_found)} out of {len(section_coords_df)} sections not found")
print(not_found["section"].tolist())
```

```python
retry_coords_v2 = []
for i, section in enumerate(not_found["section"], start=1):
    query = f"{section}, Telangana, India"
    location = geocode(query)
    if location:
        print(f"[{i}] {section} -> FOUND ({location.latitude}, {location.longitude}) | {location.address}")
        retry_coords_v2.append({"section": section, "latitude": location.latitude, "longitude": location.longitude})
    else:
        print(f"[{i}] {section} -> still not found")
        retry_coords_v2.append({"section": section, "latitude": None, "longitude": None})

retry_df_v2 = pd.DataFrame(retry_coords_v2)
```

```python
retry_df_v2 = retry_df_v2.set_index("section")
section_coords_df = section_coords_df.set_index("section")
section_coords_df.update(retry_df_v2)   # only fills in where retry_df_v2 has real values
section_coords_df = section_coords_df.reset_index()

still_missing = section_coords_df[section_coords_df["latitude"].isna()]
print(f"{len(still_missing)} sections still missing coordinates")
print(still_missing["section"].tolist())
```

```python
#Haversine Distance Function
def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Earth's radius in km
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c
```

```python
sections_unique = df3_with_coords[["section", "latitude", "longitude"]].drop_duplicates(subset="section").reset_index(drop=True)

nearest_station_list = []

for _, row in sections_unique.iterrows():
    distances = haversine(row["latitude"], row["longitude"], gw_stations["Latitude"], gw_stations["Longitude"])
    nearest_idx = distances.idxmin()
    nearest_station_list.append({
        "section": row["section"],
        "nearest_station": gw_stations.loc[nearest_idx, "Station"],
        "distance_km": distances[nearest_idx]
    })

nearest_station_df = pd.DataFrame(nearest_station_list)
print(nearest_station_df.head(10))
print(nearest_station_df["distance_km"].describe())
```

```python
far_matches = nearest_station_df[nearest_station_df["distance_km"] > 10]  # adjust threshold as you see fit
print(f"{len(far_matches)} sections have their nearest station over 10km away")
print(far_matches)
```

```python
gw_with_section = gw_hyd.merge(nearest_station_df, left_on="Station", right_on="nearest_station", how="inner")

print(gw_with_section.shape)
print(gw_with_section[["Station", "section", "distance_km"]].drop_duplicates().head())
```

```python
gw_with_section["Data Acquisition Time"] = pd.to_datetime(gw_with_section["Data Acquisition Time"], format="%d-%m-%Y %H:%M", errors="coerce")

print("Failed to parse:", gw_with_section["Data Acquisition Time"].isna().sum())

gw_with_section["year"] = gw_with_section["Data Acquisition Time"].dt.year
gw_with_section["month"] = gw_with_section["Data Acquisition Time"].dt.month
```

```python
groundwater_monthly = gw_with_section.groupby(
    ["section", "year", "month"], as_index=False
).agg({
    "Groundwater Level Telemetry 6 Hourly (meter)": ["mean", "min", "max"]
})

groundwater_monthly.columns = ["section", "year", "month", "gwl_mean", "gwl_min", "gwl_max"]

print(groundwater_monthly.shape)
print(groundwater_monthly.head())
```

```python
final_merged = df3_with_coords.merge(
    groundwater_monthly, on=["section", "year", "month"], how="left"
)

print(final_merged.shape)
print("Missing groundwater data:", final_merged["gwl_mean"].isna().sum(), "out of", len(final_merged))
```
