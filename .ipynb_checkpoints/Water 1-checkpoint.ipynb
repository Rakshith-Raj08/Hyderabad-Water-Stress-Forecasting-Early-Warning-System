```python
import numpy as np
import pandas as pd 
```

```python
df1 = pd.read_csv("DataSets/Hyderabad Water Supplies Through Tankers S1.csv")       
df2 = pd.read_csv("DataSets/Hyderabad HWSSB Billing and Collection Data S1.csv")
df3 = pd.read_csv("DataSets/Hyderabad HMWSSB Water Connections Data S1.csv")
```

```python
#Removing upper cases and spaces
def clean_sections(df, col="section"):
    return set(df[col].astype(str).str.strip().str.upper())
```

```python
s1, s2, s3 = clean_sections(df1), clean_sections(df2), clean_sections(df3)
```

```python
# Number of columns in each dataset and the ones in common
print("Unique sections — dataset 1:", len(s1))
print("Unique sections — dataset 2:", len(s2))
print("Unique sections — dataset 3:", len(s3))
print("\nCommon to all 3:", len(s1 & s2 & s3))
```

```python
#Trying to Find the Rows which are not aviailable in other Datasets 
Diff12 = s1 - s2
Diff21 =  s2 - s1    
Diff13 = s1 - s3  
Diff31 = s3 - s1
Diff23 = s2 - s3
Diff32 = s3 - s2
```

```python
# Making a dataframe of the sections that are missing in other datasets for better readability 
def to_col(s):
    return sorted(list(s))  

diffs = {
    "In df1 not df2": to_col(Diff12),
    "In df2 not df1": to_col(Diff21),
    "In df1 not df3": to_col(Diff13),
    "In df3 not df1": to_col(Diff31),
    "In df2 not df3": to_col(Diff23),
    "In df3 not df2": to_col(Diff32),
}

max_len = max(len(v) for v in diffs.values())

# pad shorter lists with empty strings so all columns are equal length
for k, v in diffs.items():
    diffs[k] = v + [""] * (max_len - len(v))

diff_table = pd.DataFrame(diffs)
diff_table
```

```python
#Manually dropping the Pattern 1 and Pattern 2 rows
#Pattern 1 rows are rows which placeholder/unassigned rows
#Pattern 2 rows are rows which Water sources/infrastructure mixed with residential sections
```

```python
#Pattern 3 rows are being merged with the Parent localities that are orginal to them
import re

def clean_section_name(name):
    name = str(name).upper().strip()
    # cut the string at the first "(" or digit or "-OLD" or "DMA" — keep only what's before it
    name = re.split(r"\(|DMA|-OLD|\d", name)[0]
    return name.strip(" -")
```

```python
dataframes = {
    "tanker": df1,
    "billing": df2,
    "connections": df3,
}

for name, d in dataframes.items():
    d["section"] = d["section"].apply(clean_section_name)
    d = d[d["section"] != ""]          # drop junk/unassigned rows
    dataframes[name] = d
```

```python
print(dataframes["tanker"].columns)
print(dataframes["billing"].columns)
print(dataframes["connections"].columns)
```

```python
#Billings Dataframe has Wdie long columns while the other too have top long columns so changing that and also changing the naming of months in billing
month_cols = [c for c in dataframes["billing"].columns if c not in ["Year", "Division", "section", "Category"]]

melted = dataframes["billing"].melt(
    id_vars=["Year", "Division", "section", "Category"],
    value_vars=month_cols,
    var_name="month_measure",
    value_name="value"
)

melted[["month", "measure"]] = melted["month_measure"].str.split(" ", n=1, expand=True)

billing_long = melted.pivot_table(
    index=["Year", "Division", "section", "Category", "month"],
    columns="measure",
    values="value",
    aggfunc="first"
).reset_index()

dataframes["billing"] = billing_long

# --- standardize month to numeric (1-12) to match tanker/connections format ---

month_map = {
    "Jan": 1, "Feb": 2, "Mar": 3, "Apr": 4, "May": 5, "June": 6,
    "July": 7, "Aug": 8, "Sept": 9, "Oct": 10, "Nov": 11, "Dec": 12
}

dataframes["billing"]["month"] = dataframes["billing"]["month"].map(month_map)
```

```python
print(dataframes["tanker"].columns)
print(dataframes["connections"].columns)
print(dataframes["billing"].columns)
```

```python
dataframes["billing"]["month"] = dataframes["billing"]["month"].astype(float)    #Making them all of the same DataType
```

```python
print(dataframes["tanker"]["month"].dtype)
print(dataframes["connections"]["month"].dtype)
print(dataframes["billing"]["month"].dtype)
```

```python
#Doing groupby to all the 3 dataframes on the basis of Year Month and Section
dataframes["tanker"] = dataframes["tanker"].groupby(
    ["year", "month", "section"], as_index=False
).agg({
    "division": "first",
    "noofbookings": "sum",
    "delivered": "sum"
})

dataframes["connections"] = dataframes["connections"].groupby(
    ["year", "month", "section"], as_index=False
).agg({
    "division": "first",
    "category": "first",
    "applied": "sum",
    "approved": "sum"
})

dataframes["billing"] = dataframes["billing"].groupby(
    ["Year", "month", "section"], as_index=False
).agg({
    "Division": "first",
    "Category": "first",
    "collection": "sum",
    "demand": "sum",
    "noofcans": "sum"
})
```

```python
# group by section AND month (drop year and category, since billing/connections only have 1 year each anyway)
connections_by_month = dataframes["connections"].groupby(
    ["section", "month"], as_index=False
).agg({
    "division": "first",
    "applied": "sum",
    "approved": "sum"
})

billing_by_month = dataframes["billing"].groupby(
    ["section", "month"], as_index=False
).agg({
    "Division": "first",
    "collection": "sum",
    "demand": "sum",
    "noofcans": "sum"
})

# merge onto tanker using ONLY section — this is what broadcasts/duplicates the value across all years
merged = dataframes["tanker"].merge(connections_by_month, on=["section", "month"], how="left")
merged = merged.merge(billing_by_month, on=["section", "month"], how="left")
```

```python
merged.shape
merged.isna().sum()
```

```python
tanker_sections = set(dataframes["tanker"]["section"])
conn_sections = set(connections_by_month["section"])
billing_sections = set(billing_by_month["section"])

print("In tanker not in connections:", tanker_sections - conn_sections)
print("In tanker not in billing:", tanker_sections - billing_sections)
```

```python
# only compare rows where all three division columns actually have values (not NaN)
non_null = merged.dropna(subset=["division_x", "division_y", "Division"])

real_mismatch = non_null[
    (non_null["division_x"] != non_null["division_y"]) |
    (non_null["division_x"] != non_null["Division"])
]

print("Total rows:", len(merged))
print("Rows with all divisions present:", len(non_null))
print("Genuine mismatches (excluding NaN rows):", len(real_mismatch))
```

```python
mismatch_examples = real_mismatch[["section", "division_x", "division_y", "Division"]].drop_duplicates()
print(mismatch_examples.head(20))
print("\nHow many unique sections are affected:", mismatch_examples["section"].nunique())
```
